## Ejemplos de dicccionario

```python
# --- PROGRAMA DE GESTIÓN DE JOYERÍA ---

# Diccionario inicial vacío
Diccionario_joyero = {}

# --- FUNCIONES ---

def agregar_joyero():
    id_joya = input("Ingrese el ID o código de la joya: ")
    material = input("Material (Oro/Plata): ")
    engaste = input("Técnica de engaste: ")
    valor = input("Valor de inventario: ")
    certificado = input("¿Tiene certificado de autenticidad? (Sí/No): ")
    diseno = input("¿Es diseño propio? (Sí/No): ")
    tienda = input("Nombre de la tienda: ")

    # Se guarda como un diccionario dentro del diccionario principal
    Diccionario_joyero[id_joya] = {
        "Material": material,
        "Engaste": engaste,
        "Valor": valor,
        "Certificado": certificado,
        "Diseño": diseno,
        "Tienda": tienda
    }
    print("\n--- Registro agregado con éxito ---")

def listar_joyero():
    if not Diccionario_joyero:
        print("\nEl inventario está vacío.")
    else:
        print(f"\n{'ID':<10} | {'Material':<10} | {'Engaste':<15} | {'Valor':<10} | {'Certificado':<12} | {'Diseño':<10} | {'Tienda':<15}")
        print("-" * 90)
        for clave, valor in Diccionario_joyero.items():
            print(f"{clave:<10} | {valor['Material']:<10} | {valor['Engaste']:<15} | {valor['Valor']:<10} | {valor['Certificado']:<12} | {valor['Diseño']:<10} | {valor['Tienda']:<15}")

def buscar_joyero():
    busqueda = input("Ingrese el ID de la joya a buscar: ")
    if busqueda in Diccionario_joyero:
        datos = Diccionario_joyero[busqueda]
        print(f"\nResultado de búsqueda para ID: {busqueda}")
        print(f"{'ID':<10} | {'Material':<10} | {'Engaste':<15} | {'Valor':<10} | {'Certificado':<12} | {'Diseño':<10} | {'Tienda':<15}")
        print("-" * 90)
        print(f"{busqueda:<10} | {datos['Material']:<10} | {datos['Engaste']:<15} | {datos['Valor']:<10} | {datos['Certificado']:<12} | {datos['Diseño']:<10} | {datos['Tienda']:<15}")
    else:
        print("\nNo se encontró ninguna joya con ese ID.")

def borrar_joyero():
    id_borrar = input("Ingrese el ID de la joya que desea eliminar: ")
    if id_borrar in Diccionario_joyero:
        del Diccionario_joyero[id_borrar]
        print(f"\nID {id_borrar} eliminado correctamente.")
    else:
        print("\nEl ID no existe.")

def actualizar_joyero():
    id_act = input("Ingrese el ID de la joya a actualizar: ")
    if id_act in Diccionario_joyero:
        print("Ingrese los nuevos datos:")
        Diccionario_joyero[id_act]["Material"] = input("Nuevo Material: ")
        Diccionario_joyero[id_act]["Engaste"] = input("Nueva Técnica de engaste: ")
        Diccionario_joyero[id_act]["Valor"] = input("Nuevo Valor: ")
        Diccionario_joyero[id_act]["Certificado"] = input("Nuevo Certificado (Sí/No): ")
        Diccionario_joyero[id_act]["Diseño"] = input("Nuevo Diseño (Sí/No): ")
        Diccionario_joyero[id_act]["Tienda"] = input("Nueva Tienda: ")
        print("\n--- Registro actualizado ---")
    else:
        print("\nEse ID no existe en el sistema.")

# --- MENÚ INTERACTIVO ---

while True:
    print("\n--- MENÚ DE GESTIÓN JOYERO ---")
    print("1. Agregar joya")
    print("2. Listar inventario")
    print("3. Buscar joya")
    print("4. Borrar joya")
    print("5. Actualizar joya")
    print("6. Salir")

    opcion = input("Seleccione una opción (1-6): ")

    if opcion == "1":
        agregar_joyero()
    elif opcion == "2":
        listar_joyero()
    elif opcion == "3":
        buscar_joyero()
    elif opcion == "4":
        borrar_joyero()
    elif opcion == "5":
        actualizar_joyero()
    elif opcion == "6":
        print("Saliendo del programa... ¡Hasta pronto!")
        break
    else:
        print("Opción no válida, intente de nuevo.")
```
